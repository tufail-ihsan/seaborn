**Bar Plot**

A bar plot shows the average (or another statistic) of a numerical variable for each category.

**Input**


One categorical variable (x)


One numerical variable (y)

**For example:**

Suppose the average age of passengers in 1st class is 38 years.

A 95% confidence interval might be:

38 ± 3

or

35 to 41 years

**What is a Confidence Interval?**

A confidence interval gives a range in which the true population mean is likely to lie.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset('tips')
```

```python
plt.style.use('fivethirtyeight')
```

```python
tips.head()
```

```python
sns.catplot(x='smoker',y='total_bill',kind='bar',data=tips)
```

```python
sns.catplot(x='smoker',y='total_bill',hue='sex',kind='bar',data=tips)
```

for another statistics

```python
# for median
sns.catplot(x='smoker',y='total_bill',hue='sex',estimator=np.median,kind='bar',data=tips)
```

```python
# for std
sns.catplot(x='smoker',y='total_bill',hue='sex',estimator=np.std,kind='bar',data=tips)
```

```python
# for var
sns.catplot(x='smoker',y='total_bill',hue='sex',estimator=np.var,kind='bar',data=tips)
```

**Count plot**

A count plot shows how many times each category appears in your dataset.

**Input**


-->Only categorical variable


-->No numerical variable needed

```python
sns.catplot(x='sex',kind='count',data=tips)
```

```python
sns.catplot(x='sex',hue='smoker',kind='count',data=tips)
```
