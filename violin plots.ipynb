**Violin plots**

A violin plot is a combination of a box plot and a density plot. It shows the distribution of numerical data across different categories, including the median, spread, and the shape of the data distribution. The wider parts of the violin indicate where more data points are concentrated, while the narrower parts show where fewer data points exist.

```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use('fivethirtyeight')
```

**load data**

```python
tips=sns.load_dataset('tips')
```

**Sample of data**


```python
tips.sample(3)
```

```python
sns.violinplot(tips['total_bill'])
```

we can use catplot also

```python
sns.catplot(x='day',y='total_bill',kind='violin',data=tips)
```

with hue parameter


```python
sns.catplot(x='day',y='tip',kind='violin',hue='sex',data=tips)
```

when we want to show it in one graph than we will use this code

```python
sns.catplot(x='day',y='tip',kind='violin',hue='sex',split='Ture',data=tips)
```
