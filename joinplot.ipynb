**joint plot**

A Joint Plot is a Seaborn visualization that shows the relationship between two numerical variables along with the distribution of each variable.

**Different Types**

1. Scatter (default)

2. Hex Plot

Useful when there are many points.

3. KDE Plot

Shows density instead of points.

4. Regression Plot

Shows a best-fit line.

**When to Use Joint Plot**

Use it when:

Both variables are numerical

**Avoid it when:**

One or both variables are categorical (Gender, City, Department, etc.)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
```

```python
tips.head()
```

1.scatter plot


```python
sns.jointplot(x='total_bill',y='tip',data=tips)
```

2.Hexbin plots

```python
sns.jointplot(x='total_bill',y='tip',kind='hex',data=tips) 
# 2D histogram
```

3. kde plot

```python
sns.jointplot(x='total_bill',y='tip',kind='kde',data=tips)
```

4. regrission plot

```python
sns.jointplot(x='total_bill',y='tip',kind='reg',data=tips)
```
