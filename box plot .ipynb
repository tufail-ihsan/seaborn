**Boxplot**

A box plot is a graph used to show the distribution of numerical data. It displays the minimum value, first quartile (Q1), median, third quartile (Q3), and maximum value, helping you understand the data's center, spread, and possible outliers. The line inside the box represents the median, the box shows where the middle 50% of the data lies, and the whiskers extend to the remaining data points. Box plots are very useful in EDA (Exploratory Data Analysis) for comparing distributions of a numerical variable across different categories and for quickly detecting outliers and skewness in the data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tip=sns.load_dataset('tips')
```

```python
tip.head()
```

```python
#univarite 
sns.boxplot(tip['total_bill'])
```

```python
sns.catplot(x='day',y='tip',kind='box',data=tip)
```

```python
sns.catplot(x='day',y='tip',hue='sex',kind='box',data=tip)
```
