**Distplot**

Distplot are used to do univatiate analysis on a continous/numerical data.

Displot combines 3 things in on graph:

1. Histogram

2. Kernal density estimation(PDF)

3. Rugplot

```python
import pandas as pd
import seaborn as sns
```

```python
titanic=sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
titanic['age'].fillna(titanic['age'].mean(), inplace= True)
```

**plotting a distplot**

```python
sns.distplot(titanic['age'])
```

**plotting only histogram**

```python
sns.displot(titanic['age'],kde=False)
```

**plotting only KDE**

```python
sns.displot(titanic['age'], kind='kde')
```

```python
sns.distplot(titanic['age'], hist=False, kde=False, rug=True)
```

```python
sns.distplot(titanic[titanic['survived']==1]['age'])
sns.distplot(titanic[titanic['survived']==0]['age'])
```
