```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.use('fivethirtyeight')
```

```python
tips=sns.load_dataset('tips')
```

```python
tips.head()
```

**scatter plot**

numerical --> numerical plot

```python
sns.scatterplot(x='total_bill',y='tip',hue='sex',style='smoker',size='size',data=tips)
#hue= add colur based on categorical variable
#style= change the shape of markes. useful when colur is not enough
#size= change the size of point but this only work on numerica colum
```

**categorical scatter plot**

types: a) strip b) swarm plot

numerical--->categorical

x-axis= categorical

y-axis= numerical

**a)strip plot**

```python
sns.catplot(x='day',y='tip',kind='strip',data=tips,palette='Set2')
#by defult it is strip
```

```python
# OR
sns.stripplot(x='day',y='tip',data=tips,palette='Set1')
```

```python
# with hue parameter
sns.stripplot(x='day',y='tip',hue='sex',data=tips,palette='viridis')
```

**b) swarm plot**

```python
sns.catplot(x='day',y='tip',kind='swarm',data=tips,palette='Set1')
#or 
#sns.swarmplot(x='day',y='tip',data=tips,palette='Set1')
```

with using hue parameter

```python
# with hue parameter
sns.swarmplot(x='day',y='tip',hue='sex',data=tips,palette='viridis')
```
