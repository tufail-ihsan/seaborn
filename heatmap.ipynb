**Heat map**

A heatmap is a visualization that uses colors to represent values in a table or matrix.

-> Darker/lighter colors indicate higher or lower values.


-> It helps you quickly find patterns, relationships, and strong correlations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights=sns.load_dataset('flights')
```

```python
flights.head()
```

**Making pivot table for heatmap**

```python
x=flights.pivot_table(index='year',columns='month',values='passengers',aggfunc='sum')
```

```python
x
```

```python
sns.heatmap(x,cbar=False)
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5)
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5,annot=True,fmt='d')
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5,annot=True,fmt='d',cmap='viridis')
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5,annot=True,fmt='d',cmap='winter')
```

**clustermap**

It not only shows values using colors, but also groups similar rows and columns together automatically.

```python
sns.clustermap(x)
```

```python
sns.clustermap(x, metric='correlation')
```

```python
sns.clustermap(x,z_score=0,annot=True ,metric='correlation')
```

```python
sns.clustermap(x,z_score=0,annot=True,row_cluster=False,metric='correlation')
```
